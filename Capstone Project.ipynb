## Introduction

The Vehicle Rental and Fleet Management System is a Python-based application designed to manage vehicle rentals and fleet operations. The system allows users to register customers, add different types of vehicles, search for available vehicles, rent and return vehicles, calculate rental charges and process payments.

The project uses Object-Oriented Programming (OOP) concepts such as inheritance, encapsulation, polymorphism and method overriding. It also incorporates exception handling to manage errors and ensure that the system operates correctly.

```python
from datetime import date, datetime, timedelta
```

## Importing Required Libraries

In this section, we import the necessary Python libraries for the project. The datetime library will be used to record rental dates and return dates. These libraries help us organize and manage rental information.

```python
class Vehicle:

    def __init__(self, vehicle_id, model, registration_number, daily_rate):
        self.vehicle_id = vehicle_id
        self.model = model
        self.registration_number = registration_number
        self.daily_rate = daily_rate
        self.available = True

    def display_information(self):
        print("Vehicle ID:", self.vehicle_id)
        print("Model:", self.model)
        print("Registration Number:", self.registration_number)
        print("Daily Rate:", self.daily_rate)
        print("Availability:", "Available" if self.available else "Rented")

    def update_availability(self, status):
        self.available = status

    def calculate_rental_cost(self, days):
        if days <= 0:
            raise ValueError("Rental period must be greater than zero.")

        return self.daily_rate * days
```

## Creating the Vehicle Base Class

The Vehicle class is the parent class of the different vehicle types in the system. It contains common attributes such as vehicle ID, model, registration number, rental rate and availability status.

It also contains methods for displaying vehicle information, checking availability and updating vehicle availability. Other vehicle classes will inherit these attributes and methods.

```python
class Car(Vehicle):

    def calculate_rental_cost(self, days):
        if days <= 0:
            raise ValueError("Rental period must be greater than zero.")

        return (self.daily_rate * days) + 20


class Motorcycle(Vehicle):

    def calculate_rental_cost(self, days):
        if days <= 0:
            raise ValueError("Rental period must be greater than zero.")

        return (self.daily_rate * days) + 5


class Van(Vehicle):

    def calculate_rental_cost(self, days):
        if days <= 0:
            raise ValueError("Rental period must be greater than zero.")

        return (self.daily_rate * days) + 50


class Bus(Vehicle):

    def calculate_rental_cost(self, days):
        if days <= 0:
            raise ValueError("Rental period must be greater than zero.")

        return (self.daily_rate * days) + 100
```

## Creating Different Vehicle Types

In this section, we create four vehicle subclasses: Car, Motorcycle, Van and Bus.

Each class inherits from the Vehicle parent class. They also override the calculate_rental_cost() method to calculate rental charges differently.

This demonstrates inheritance, polymorphism and method overriding in Python.
Each vehicle type has a different additional rental fee. For example, a car adds 20, a motorcycle adds 5, a van adds 50 and a bus adds 100 to the basic rental charge.

The same method name produces different results depending on the vehicle object. This is polymorphism.

```python
class Person:

    def __init__(self, name, phone, email):
        self.name = name
        self.phone = phone
        self.email = email


class Customer(Person):

    def __init__(self, customer_id, name, phone, email):
        super().__init__(name, phone, email)
        self.customer_id = customer_id

    def display_customer(self):
        print("Customer ID:", self.customer_id)
        print("Name:", self.name)
        print("Phone:", self.phone)
        print("Email:", self.email)
```

## Creating the Person and Customer Classes

The Person class stores general personal information such as name, phone number and email address.

The Customer class inherits from Person and adds a customer ID. This demonstrates inheritance because the Customer class reuses the attributes and methods of the Person class.

```python
class Rental:

    def __init__(self, rental_id, customer_id, vehicle_id, rental_days, total_cost):

        self.rental_id = rental_id
        self.customer_id = customer_id
        self.vehicle_id = vehicle_id
        self.rental_days = rental_days
        self.total_cost = total_cost

        self.start_date = date.today()
        self.due_date = self.start_date + timedelta(days=rental_days)

        self.status = "Active"
        self.payment_status = "Pending"

    def display_rental(self):

        print("\n----- RENTAL DETAILS -----")
        print("Rental ID:", self.rental_id)
        print("Customer ID:", self.customer_id)
        print("Vehicle ID:", self.vehicle_id)
        print("Rental Days:", self.rental_days)
        print("Total Cost:", self.total_cost)
        print("Start Date:", self.start_date)
        print("Due Date:", self.due_date)
        print("Rental Status:", self.status)
        print("Payment Status:", self.payment_status)

   
```

## Creating the Rental Class

The Rental class represents a vehicle rental transaction. It stores the rental ID, customer, vehicle, rental duration, rental date, total rental cost, and rental status.

The class calculates the rental charges using the calculate_rental_cost() method from the selected vehicle object. It also provides a method for displaying rental details.

```python
class Payment:

    payment_counter = 1

    def __init__(self, rental_id, amount_due):

        self.payment_id = f"P{Payment.payment_counter:03d}"
        Payment.payment_counter += 1

        self.rental_id = rental_id
        self.amount_due = amount_due
        self.amount_paid = 0
        self.balance = amount_due
        self.change = 0
        self.status = "Pending"
        self.payment_history = []

    def process_payment(self, amount, overdue=False):

        if amount <= 0:
            print("Enter a valid payment amount.")
            return

        if self.status == "Paid":
            print("This rental is already fully paid.")
            return

        self.change = 0

        remaining = self.balance

        if amount > remaining:
            self.change = amount - remaining
            amount = remaining

        self.amount_paid += amount
        self.balance = self.amount_due - self.amount_paid

        payment_time = datetime.now()

        self.payment_history.append({
            "amount": amount,
            "date_time": payment_time
        })

        if self.balance == 0:
            self.status = "Paid"
        elif overdue:
            self.status = "Overdue"
        else:
            self.status = "Pending"

        print("\n----- PAYMENT DETAILS -----")
        print("Payment ID:", self.payment_id)
        print("Rental ID:", self.rental_id)
        print("Amount Due:", self.amount_due)
        print("Amount Paid:", self.amount_paid)
        print("Outstanding Balance:", self.balance)

        if overdue:
            print("NOTICE: This payment is overdue.")
       
        if self.change > 0:
            print("Excess Payment:", self.change)
            print("Change to Customer:", self.change)

        if self.balance > 0:
            print("INSUFFICIENT PAYMENT.")
        else:
            print("Payment completed successfully.")

        print("Payment Status:", self.status)
        print("Payment Date and Time:", payment_time.strftime("%d-%m-%Y %H:%M:%S"))

    def display_payment(self):

        print("\n----- PAYMENT HISTORY -----")
        print("Payment ID:", self.payment_id)
        print("Rental ID:", self.rental_id)

        for payment in self.payment_history:
            print("Amount:", payment["amount"])
            print("Date and Time:",
                  payment["date_time"].strftime("%d-%m-%Y %H:%M:%S"))

        print("Total Paid:", self.amount_paid)
        print("Outstanding Balance:", self.balance)
        print("Payment Status:", self.status)
```

## Creating the Payment Class

The Payment class manages payments made by customers for their vehicle rentals.

It stores the rental information, amount paid and payment status. It also checks if the amount is paid in full or if there is outsanding balance.If excess payment is made , the balance . It also checks whether the payment amount is valid and whether it covers the total rental charges.

```python
class RentalSystem:

    def __init__(self):
        self.customers = {}
        self.vehicles = {}
        self.rentals = {}
        self.payments = {}

    # 1. Register Customer
    def register_customer(self, customer_id, name, phone, email):

        if customer_id in self.customers:
            print("Customer ID already exists.")
            return

        customer = Customer(customer_id, name, phone, email)
        self.customers[customer_id] = customer

        print("Customer registered successfully.")

    # 2. Add Vehicle
    def add_vehicle(self, vehicle):

        if vehicle.vehicle_id in self.vehicles:
            print("Vehicle ID already exists.")
            return

        self.vehicles[vehicle.vehicle_id] = vehicle

        print("Vehicle added successfully.")

    # 3. View All Vehicles
    def view_all_vehicles(self):

        if not self.vehicles:
            print("No vehicles registered.")
            return

        for vehicle in self.vehicles.values():
            vehicle.display_information()
            print("-------------------------")

    # 4. View Available Vehicles
    def view_available_vehicles(self):

        found = False

        for vehicle in self.vehicles.values():

            if vehicle.available:
                vehicle.display_information()
                print("-------------------------")
                found = True

        if not found:
            print("No vehicles are currently available.")

    # 5. Search Vehicle
    def search_vehicle(self, vehicle_id):

        vehicle = self.vehicles.get(vehicle_id)

        if vehicle is None:
            print("Vehicle not found.")
        else:
            vehicle.display_information()

    # 6. Rent Vehicle
    def rent_vehicle(self, rental_id, customer_id, vehicle_id, days):

        if days <= 0:
            print("Invalid rental period.")
            return

        customer = self.customers.get(customer_id)
        vehicle = self.vehicles.get(vehicle_id)

        if customer is None:
            print("Customer not found.")
            return

        if vehicle is None:
            print("Vehicle not found.")
            return

        if not vehicle.available:
            print("Vehicle is unavailable.")
            return

        if rental_id in self.rentals:
            print("Rental ID already exists.")
            return

        try:
            total_cost = vehicle.calculate_rental_cost(days)
            rental = Rental(rental_id, customer_id, vehicle_id, days,total_cost)

            self.rentals[rental_id] = rental
            vehicle.update_availability(False)

            print("Vehicle rented successfully.")
            rental.display_rental()

        except ValueError as error:
            print("Error:", error)

    # 7. Return Vehicle
    def return_vehicle(self, rental_id):

        rental = self.rentals.get(rental_id)

        if rental is None or rental.status != "Active":
            print("Rental not found or vehicle has already been returned.")
            return


        rental.status = "Returned"
        vehicle = self.vehicles.get(rental.vehicle_id)
        if vehicle:
            vehicle.update_availability(True)
    

        print("Vehicle returned successfully.")
        print("Vehicle is now available.")

    # 8. Calculate Rental Cost
    def calculate_rental_cost(self, vehicle_id, days):

        vehicle = self.vehicles.get(vehicle_id)

        if vehicle is None:
            print("Vehicle not found.")
            return

        try:
            cost = vehicle.calculate_rental_cost(days)
            print("Total Rental Cost:", cost)

        except ValueError as error:
            print("Error:", error)

    # 9. Process Payment
    def process_payment(self, rental_id, amount):

        rental = self.rentals.get(rental_id)

        if rental is None:
            print("Rental not found.")
            return



        if rental_id not in self.payments:
            self.payments[rental_id] = Payment(
                rental_id, rental.total_cost
             )

        payment = self.payments[rental_id]

        if payment.status == "Paid":
            print("Rental has already been fully paid.")
            return

        due_date = getattr(rental, "due_date", None)
        overdue = False

        if due_date:
             if isinstance(due_date, str):
                 due_date = date.fromisoformat(due_date)

             if isinstance(due_date, datetime):
                 due_date = due_date.date()

             overdue = date.today() > due_date

        payment.process_payment(amount, overdue)

        rental.payment_status = payment.status

    # 10. Customer Rental History
    def customer_rental_history(self, customer_id):

        customer = self.customers.get(customer_id)

        if customer is None:
            print("Customer not found.")
            return

        print("Rental History for:", customer.name)

        found = False

        for rental in self.rentals.values():

            if rental.customer_id == customer_id:
                rental.display_rental()
                print("-------------------------")
                found = True

        if not found:
            print("No rental history found.")

    # 11. View Vehicle Information
    def view_vehicle_information(self, vehicle_id):

        vehicle = self.vehicles.get(vehicle_id)

        if vehicle is None:
            print("Vehicle not found.")
            return

        vehicle.display_information()

    # 12. Update Vehicle Availability

    def update_vehicle_availability(self, vehicle_id, is_available):

        if vehicle_id not in self.vehicles:
            print("Error: Vehicle not found.")
            return

        vehicle = self.vehicles[vehicle_id]

        # Allow unavailable status even if vehicle is not rented
        if not is_available:
            vehicle.update_availability(False)
            print(f"Vehicle {vehicle_id} is now unavailable.")
            print("Reason may include maintenance or mechanical problems.")
            return

        # Prevent a rented vehicle from being marked available
        if any(
            rental.vehicle_id == vehicle_id and rental.status == "Active"
            for rental in self.rentals.values()
        ):
            print("Error: Vehicle is currently rented.")
            print("It cannot be marked available until returned.")
            return

        vehicle.update_availability(True)
        print(f"Vehicle {vehicle_id} is now available.")


    # 13. Display Fleet Summary
    def display_fleet_summary(self):

        total = len(self.vehicles)

        available = 0
        rented = 0

        for vehicle in self.vehicles.values():

            if vehicle.available:
                available += 1
            else:
                rented += 1

        print("========== FLEET SUMMARY ==========")
        print("Total Vehicles:", total)
        print("Available Vehicles:", available)
        print("Rented Vehicles:", rented)
        print("Total Customers:", len(self.customers))
        print("Total Rentals:", len(self.rentals))
        print("Total Payments:", len(self.payments))
        print("===================================")
```

```python
def main():

    system = RentalSystem()

    while True:

        print("\n========================================")
        print(" VEHICLE RENTAL MANAGEMENT SYSTEM")
        print("========================================")
        print("1. Register Customer")
        print("2. Add Vehicle")
        print("3. View All Vehicles")
        print("4. View Available Vehicles")
        print("5. Search Vehicle")
        print("6. Rent Vehicle")
        print("7. Return Vehicle")
        print("8. Calculate Rental Cost")
        print("9. Process Payment")
        print("10. View Customer Rental History")
        print("11. View Vehicle Information")
        print("12. Update Vehicle Availability")
        print("13. Display Fleet Summary")
        print("14. Exit")
        print("========================================")

        try:
            choice = int(input("Enter your choice: "))

            # Option 1: Register Customer
            if choice == 1:

                customer_id = input("Enter Customer ID: ")
                name = input("Enter Customer Name: ")
                phone = input("Enter Phone Number: ")
                email = input("Enter Email: ")

                system.register_customer(
                    customer_id, name, phone, email
                )

            # Option 2: Add Vehicle
            elif choice == 2:

                vehicle_id = input("Enter Vehicle ID: ")
                model = input("Enter Vehicle Model: ")
                registration = input("Enter Registration Number: ")
                rate = float(input("Enter Daily Rental Rate: "))

                if rate <= 0:
                    print("Rental rate must be greater than zero.")
                    continue

                print("\nSelect Vehicle Type:")
                print("1. Car")
                print("2. Motorcycle")
                print("3. Van")
                print("4. Bus")

                vehicle_type = input("Enter vehicle type: ")

                if vehicle_type == "1":
                    vehicle = Car(
                        vehicle_id, model, registration, rate
                    )

                elif vehicle_type == "2":
                    vehicle = Motorcycle(
                        vehicle_id, model, registration, rate
                    )

                elif vehicle_type == "3":
                    vehicle = Van(
                        vehicle_id, model, registration, rate
                    )

                elif vehicle_type == "4":
                    vehicle = Bus(
                        vehicle_id, model, registration, rate
                    )

                else:
                    print("Invalid vehicle type.")
                    continue

                system.add_vehicle(vehicle)

            # Option 3: View All Vehicles
            elif choice == 3:
                system.view_all_vehicles()

            # Option 4: View Available Vehicles
            elif choice == 4:
                system.view_available_vehicles()

            # Option 5: Search Vehicle
            elif choice == 5:

                vehicle_id = input("Enter Vehicle ID: ")
                system.search_vehicle(vehicle_id)

            # Option 6: Rent Vehicle
            elif choice == 6:

                rental_id = input("Enter Rental ID: ")
                customer_id = input("Enter Customer ID: ")
                vehicle_id = input("Enter Vehicle ID: ")
                days = int(input("Enter Number of Rental Days: "))

                system.rent_vehicle(
                    rental_id, customer_id, vehicle_id, days
                )

            # Option 7: Return Vehicle
            elif choice == 7:

                rental_id = input("Enter Rental ID: ")
                system.return_vehicle(rental_id)

            # Option 8: Calculate Rental Cost
            elif choice == 8:

                vehicle_id = input("Enter Vehicle ID: ")
                days = int(input("Enter Number of Days: "))

                system.calculate_rental_cost(vehicle_id, days)

        
            # Option 9: Process Payment
            elif choice == 9:

                rental_id = input("Enter Rental ID: ")

                try:
                    amount = float(input("Enter Payment Amount: "))

                    system.process_payment(rental_id, amount)

                except ValueError:
                    print("Please enter a valid number.")
        

            # Option 10: Customer Rental History
            elif choice == 10:

                customer_id = input("Enter Customer ID: ")
                system.customer_rental_history(customer_id)

            # Option 11: View Vehicle Information
            elif choice == 11:

                vehicle_id = input("Enter Vehicle ID: ")
                system.view_vehicle_information(vehicle_id)

            # Option 12: Update Vehicle Availability
            elif choice == 12:

                vehicle_id = input("Enter Vehicle ID: ")

                print("1. Available")
                print("2. Unavailable")

                status_choice = input("Select new status: ")

                if status_choice == "1":
                    status = True

                elif status_choice == "2":
                    status = False

                else:
                    print("Invalid availability option.")
                    continue

                system.update_vehicle_availability(
                    vehicle_id, status
                )

            # Option 13: Display Fleet Summary
            elif choice == 13:

                system.display_fleet_summary()

            # Option 14: Exit
            elif choice == 14:

                print("Thank you for using the Vehicle Rental System.")
                break

            else:
                print("Invalid choice. Please select between 1 and 14.")

        except ValueError:
            print("Invalid input. Please enter the correct data type.")

        except Exception as error:
            print("An unexpected error occurred:", error)
```
